#### Cơ cấu nguồn vốn

```python
years = np.array(list(range(2015, 2024)))
x = np.linspace(years[0] - 0.5, years[-1] + 0.5, 1000)
colors_debt = '#E58D36'
colors_equity = '#438AAC'
debt_ratios = []
equity_ratios = []

for year_idx in range(len(years)):
    total_debt_sum = 0
    total_equity_sum = 0
    total_capital_sum = 0
    valid_datasets = 0

    for data in datasets:
        try:
            df = data[year_idx]
            total_debt = df['I. NỢ PHẢI TRẢ'].sum()
            equity = df['II. VỐN CHỦ SỞ HỮU'].sum()
            total_capital = df['B. TỔNG CỘNG NGUỒN VỐN'].sum()

            total_debt_sum += total_debt
            total_equity_sum += equity
            total_capital_sum += total_capital
            valid_datasets += 1
        except (KeyError, AttributeError, IndexError) as e:
            print(f"Lỗi khi xử lý dữ liệu cho năm {years[year_idx]}: {e}")
            continue

    if total_capital_sum == 0 or valid_datasets == 0:
        debt_ratio = 0
        equity_ratio = 0
    else:
        debt_ratio = (total_debt_sum / total_capital_sum) * 100
        equity_ratio = (total_equity_sum / total_capital_sum) * 100

    debt_ratios.append(debt_ratio)
    equity_ratios.append(equity_ratio)

from scipy.interpolate import interp1d
debt_interp = interp1d(years, debt_ratios, kind='linear', fill_value='extrapolate')
equity_interp = interp1d(years, equity_ratios, kind='linear', fill_value='extrapolate')

debt_values = debt_interp(x)
equity_values = equity_interp(x)

plt.figure(figsize=(14, 6))

plt.fill_between(x, 0, debt_values, color=colors_debt, label='Nợ phải trả')

plt.fill_between(x, debt_values, debt_values + equity_values, color=colors_equity, label='Vốn chủ sở hữu')

for i, year in enumerate(years):
    debt_value = debt_ratios[i]
    equity_value = equity_ratios[i]

    debt_label_y = debt_value / 2  # Nhãn nợ ở giữa vùng nợ
    equity_label_y = debt_value + (equity_value / 2)

    plt.text(year, debt_label_y, f'{debt_value:.1f}%',
             ha='center', va='center', fontsize=9, color='white', fontweight='bold')

    plt.text(year, equity_label_y, f'{equity_value:.1f}%',
             ha='center', va='center', fontsize=9, color='black', fontweight='bold')
plt.xticks(years)
plt.xlabel('Năm', fontsize=12)
plt.ylabel('Tỷ lệ (%)', fontsize=12)
plt.title('Tỷ trọng cơ cấu nguồn vốn trung bình của các ngành', fontsize=14, pad=20)
plt.ylim(0, 110)
plt.xlim(x[0], x[-1])
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.legend(ncol=2, fontsize=9, loc='upper center', bbox_to_anchor=(0.5, -0.15))
plt.tight_layout()
plt.savefig('average_capital_structure_area_fullwidth_labels.png')
plt.show()
```

#### Vòng quay tài sản và hệ số nợ

```python
# PHÂN TÍCH VÒNG QUAY TỔNG TÀI SẢN & HỆ SỐ NỢ
# Tính trung bình theo năm cho toàn bộ dữ liệu
df_at = df.groupby("Year")[["Vòng quay tổng tài sản"]].mean()
df["D/E"] = df["D/E"].replace([float('inf'), float('-inf')], float('nan'))
df_de = df.groupby("Year")[["D/E"]].mean()

# Trục chính
ax1 = plt.gca()
ax1.plot(df_at.index, df_at["Vòng quay tổng tài sản"], marker="o", color="#0073CF", label="Vòng quay tổng tài sản")
ax1.set_xlabel("Year")
ax1.set_ylabel("Vòng quay tổng tài sản", color="#0073CF")
ax1.tick_params(axis='y', labelcolor="#0073CF")

# Trục phụ
ax2 = ax1.twinx()
ax2.plot(df_de.index, df_de["D/E"], marker="s", linestyle="--", color="#B22222", label="D/E")
ax2.set_ylabel("D/E", color="#B22222")
ax2.tick_params(axis='y', labelcolor="#B22222")

# Tiêu đề và lưới
plt.title("Vòng quay tổng tài sản và D/E trung bình theo năm")
ax1.grid(True, linestyle='--', alpha=0.7)
# Thêm chú thích cho cả hai đường
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
plt.legend(lines1 + lines2, labels1 + labels2, loc="upper right")

plt.show()
```

#### Hệ số thanh toán

```python
# PHÂN TÍCH CÁC HỆ SỐ THANH TOÁN
# Chọn các cột cần thiết
cols = ['Year', 'Hệ số thanh toán ngắn hạn', 'Hệ số thanh toán nhanh', 'Khả năng thanh toán hiện hành']
df_filtered = data[cols].dropna()

# Đảm bảo cột Year là kiểu số
df_filtered['Year'] = pd.to_numeric(df_filtered['Year'], errors='coerce')

# Tính giá trị trung bình theo từng năm
mean_by_year = df_filtered.groupby('Year').mean().reset_index()

# Vẽ biểu đồ
plt.figure(figsize=(10, 6))
plt.plot(mean_by_year['Year'], mean_by_year['Hệ số thanh toán ngắn hạn'], marker='o', label='Thanh toán ngắn hạn (TB)')
plt.plot(mean_by_year['Year'], mean_by_year['Hệ số thanh toán nhanh'], marker='s', label='Thanh toán nhanh (TB)')
plt.plot(mean_by_year['Year'], mean_by_year['Khả năng thanh toán hiện hành'], marker='^', label='Thanh toán hiện hành (TB)')

plt.title('Trung bình các hệ số thanh toán theo năm')
plt.xlabel('Năm')
plt.ylabel('Giá trị trung bình')
plt.legend()
plt.grid(True)

# Đặt trục y bắt đầu từ 0 để hiện mốc 0
plt.ylim(bottom=0)

plt.tight_layout()
plt.show()
```
